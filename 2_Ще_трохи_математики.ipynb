1. Обчисліть радіус кола, знаючи його площу. Нагадаємо формулу:
$$
  S=\pi*R^2 => R =\sqrt{\frac{S}{\pi}}
$$
Для обчислення скористаємось знайомими функціями ```sqrt``` та ```pi``` з пакету ```math```.


````python
from math import sqrt,pi
S = float(input('Введіть площу: '))
r = sqrt(S/pi)
print ('Радіуc %6.3f см' %r)
````

2. А тепер за зразком знайдіть в наступній комірці площу сфери та об'єм кулі. Нагадуємо формули:
$$
  S=4*\pi*R^2    
$$
$$
  V=\frac{4}{3}*\pi*R^3
$$
Також нагадаємо, що ступінь в Python задається так: ```**```

````python
from math import pi

R = float(input('Введіть радіус кулі в сантиметрах: '))

if R < 0:
    print('Радіус не може бути відємним.')
else:
    S = 4 * pi * R**2
    V = (4/3) * pi * R**3

    print(f'Радіус: {R:.2f} см')
    print(f'Площа поверхні сфери: {S:.2f} см²')
    print(f'Обєм кулі: {V:.2f} см³')
````

3. А ще давайте переведемо температуру за Фаренгейтом у градуси Цельсія і навпаки. Формули тут надзвичайно прості:
$$
  C=(F-32)*0.5556
$$
$$
  F=C*1.8+32
$$
Спробуйте в наступній комірці! Пам'ятайте про "красиве" введення і виведення даних.

````python
first_number = int(input('Введіть 1 число: ')) ;
second_number = int(input('Введіть 2 число: ')) ;
C = (first_number - 32) * 0.5556
F = second_number * 1.8 + 32
print(f"{first_number:.1f}°Фаренгейт = {C:.1f}°Цельсій")
print(f"{second_number:.1f}°Цельсій = {F:.1f}°Фаренгейт")
````
